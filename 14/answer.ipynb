```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv('pakistani_ecommerce_dataset.csv')
df2 = pd.read_csv('student_lifestyle_dataset.csv')
```

```python
print(df1.groupby('product_category').count())
print(df1['order_value'].mean())
print(df1.groupby('city').count())
print(df1.groupby('payment_method').count())

city_avg = df1.groupby('city')['order_value'].mean()
plt.bar(city_avg.index, city_avg.values)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')
plt.show()
```

```python
print(df2['study_hours'].mean())
print(df2.groupby('gender')['sleep_hours'].mean())

plt.scatter(df2['screen_time'], df2['sleep_hours'])
plt.xlabel('Screen Time')
plt.ylabel('Sleep Hours')
plt.title('Screen Time vs Sleep Hours')
plt.show()
```

```python
plt.hist(df2['daily_transport_cost'], bins=20)
plt.xlabel('Transport Cost')
plt.ylabel('Frequency')
plt.title('Daily Transport Cost')
plt.show()
```

```python
gender_internet = df2.groupby('gender')['internet_usage'].mean()
plt.bar(gender_internet.index, gender_internet.values)
plt.xlabel('Gender')
plt.ylabel('Average Internet Usage')
plt.title('Average Internet Usage by Gender')
plt.show()
```
