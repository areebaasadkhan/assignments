```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ride_hailing_dataset.csv")
df.head()
```

```python
df.groupby("service_type")["fare"].mean()
```

```python
df["city"].value_counts()
```

```python
df["city"].value_counts().idxmax()
```

```python
df["distance_km"].mean()
```

```python
df.groupby("gender")["trip_rating"].mean()
```

```python
df.groupby("gender")["trip_rating"].agg(["mean", "count"])
```

```python
df.groupby("service_type")["fare"].mean().plot(kind="bar")
plt.xlabel("Service Type")
plt.ylabel("Average Fare")
plt.title("Average Fare by Service Type")
plt.show()
```
